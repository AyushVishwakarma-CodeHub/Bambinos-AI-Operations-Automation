```python
import pandas as pd
import numpy as np

# Load all five datasets
invoices = pd.read_csv('/content/1_invoices.csv')

faculty_forms = pd.read_csv(
    '/content/2_faculty_form_unapproved.csv'
)

booking_requests = pd.read_csv(
    '/content/3_booking_requests.csv'
)

class_bookings = pd.read_csv(
    '/content/4_class_schedule_bookings.csv'
)

students_batches = pd.read_csv(
    '/content/reference_active_batches_and_students.csv'
)

print("All datasets loaded successfully!")
```

```python
# ==========================================
# CELL 5: AUTOMATED DATA RECONCILIATION
# ==========================================

# Make copies to protect original datasets
inv = invoices.copy()
forms = faculty_forms.copy()

# Ensure IDs and amounts use consistent formats
for df, columns in [
    (inv, ['id', 'student_id']),
    (forms, ['invoice_id', 'student_id'])
]:
    for col in columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

inv['net_amount'] = pd.to_numeric(
    inv['net_amount'], errors='coerce'
)

forms['amount_paid'] = pd.to_numeric(
    forms['amount_paid'], errors='coerce'
)

# ------------------------------------------
# 1. Find duplicate renewal forms
# ------------------------------------------

duplicate_forms = forms[
    forms.duplicated(
        subset=['invoice_id'],
        keep=False
    )
]

# ------------------------------------------
# 2. Match faculty forms with invoices
# ------------------------------------------

reconciliation = forms.merge(
    inv,
    left_on='invoice_id',
    right_on='id',
    how='left',
    suffixes=('_form', '_invoice'),
    indicator=True
)

# ------------------------------------------
# 3. Check matching student IDs
# ------------------------------------------

reconciliation['student_match'] = (
    reconciliation['student_id_form']
    == reconciliation['student_id_invoice']
)

# ------------------------------------------
# 4. Check payment amount
# ------------------------------------------

reconciliation['amount_match'] = (
    reconciliation['amount_paid']
    == reconciliation['net_amount']
)

# ------------------------------------------
# 5. Generate exception reasons
# ------------------------------------------

def identify_issue(row):

    issues = []

    if row['_merge'] == 'left_only':
        issues.append('INVOICE_NOT_FOUND')

    if row['_merge'] == 'both':

        if not row['student_match']:
            issues.append('STUDENT_ID_MISMATCH')

        if row['payment_status'] != 'Paid':
            issues.append(
                'PAYMENT_' + str(row['payment_status']).upper()
            )

        if not row['amount_match']:
            issues.append('AMOUNT_MISMATCH')

    if forms['invoice_id'].value_counts().get(
        row['invoice_id'], 0
    ) > 1:
        issues.append('DUPLICATE_FORM')

    return ', '.join(issues) if issues else 'VALID'

reconciliation['reason_code'] = reconciliation.apply(
    identify_issue,
    axis=1
)

# ------------------------------------------
# 6. Display reconciliation results
# ------------------------------------------

display(
    reconciliation[
        [
            'student_id_form',
            'invoice_id',
            'net_amount',
            'amount_paid',
            'payment_status',
            'approval_status',
            'reason_code'
        ]
    ]
)

print("\nDuplicate Renewal Forms:")
display(duplicate_forms)
```

```python
# CELL 6: BATCH CAPACITY ANALYSIS

batch_analysis = students_batches[
    [
        'current_classbatch_id',
        'classbatch_name',
        'active_students_count',
        'max_batch_size'
    ]
].copy()

batch_analysis['available_seats'] = (
    batch_analysis['max_batch_size']
    - batch_analysis['active_students_count']
)

batch_analysis['capacity_status'] = batch_analysis[
    'available_seats'
].apply(
    lambda x: 'FULL' if x <= 0 else 'AVAILABLE'
)

display(batch_analysis)
```

```python
# CELL 7: CLEAN BATCH CAPACITY DATA

unique_batches = students_batches[
    [
        'current_classbatch_id',
        'classbatch_name',
        'active_students_count',
        'max_batch_size'
    ]
].drop_duplicates(
    subset=['current_classbatch_id']
).copy()

unique_batches['available_seats'] = (
    unique_batches['max_batch_size']
    - unique_batches['active_students_count']
)

unique_batches['capacity_status'] = np.where(
    unique_batches['available_seats'] > 0,
    'AVAILABLE',
    'FULL'
)

print("Total Unique Batches:", len(unique_batches))

display(unique_batches)
```

```python
# ============================================
# CELL 8: MAIN RENEWAL RECONCILIATION ENGINE
# ============================================

import pandas as pd
import numpy as np

# Create working copies
inv = invoices.copy()
forms = faculty_forms.copy()
requests = booking_requests.copy()
schedules = class_bookings.copy()
students = students_batches.copy()

# Standardize IDs
for df, cols in [
    (inv, ['id', 'student_id']),
    (forms, ['id', 'invoice_id', 'student_id']),
    (requests, ['invoice_id', 'student_id']),
    (schedules, ['invoice_id', 'student_id']),
    (students, ['student_id', 'current_classbatch_id'])
]:
    for col in cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Identify duplicate forms
duplicate_invoice_ids = set(
    forms[
        forms.duplicated('invoice_id', keep=False)
    ]['invoice_id'].dropna()
)

# Existing booking and schedule references
existing_request_invoices = set(
    requests['invoice_id'].dropna()
)

existing_schedule_invoices = set(
    schedules.loc[
        schedules['is_cancelled'].astype(str).str.lower() == 'no',
        'invoice_id'
    ].dropna()
)

# Lookup tables
student_lookup = students.drop_duplicates(
    'student_id'
).set_index('student_id')

batch_lookup = unique_batches.set_index(
    'current_classbatch_id'
)

# Process each invoice
results = []

for _, invoice in inv.iterrows():

    invoice_id = invoice['id']
    student_id = invoice['student_id']

    matching_forms = forms[
        forms['invoice_id'] == invoice_id
    ]

    issues = []
    action = "MANUAL_REVIEW"

    # Payment validation
    if invoice['payment_status'] != 'Paid':
        issues.append("PAYMENT_NOT_COMPLETED")

    # Find matching forms
    if matching_forms.empty:

        issues.append("MISSING_FACULTY_FORM")

    else:

        # Student validation
        valid_student_forms = matching_forms[
            matching_forms['student_id'] == student_id
        ]

        if valid_student_forms.empty:
            issues.append("STUDENT_ID_MISMATCH")

        if invoice_id in duplicate_invoice_ids:
            issues.append("DUPLICATE_FORM")

        # Amount validation
        for _, form in matching_forms.iterrows():

            if form['amount_paid'] != invoice['net_amount']:
                issues.append("AMOUNT_MISMATCH")
                break

    # Existing schedule validation
    if invoice_id in existing_schedule_invoices:
        issues.append("ALREADY_SCHEDULED")

    # Existing booking validation
    if invoice_id in existing_request_invoices:
        issues.append("BOOKING_REQUEST_EXISTS")

    # Check student and batch
    batch_id = None
    capacity_status = "UNKNOWN"

    if student_id in student_lookup.index:

        student = student_lookup.loc[student_id]

        batch_id = student['current_classbatch_id']

        if batch_id in batch_lookup.index:

            capacity_status = batch_lookup.loc[
                batch_id, 'capacity_status'
            ]

    # Determine final decision
    if not issues:

        if capacity_status == "AVAILABLE":
            action = "ELIGIBLE_FOR_AUTO_APPROVAL"

        elif capacity_status == "FULL":
            action = "OPS_REVIEW_BATCH_FULL"

        else:
            action = "OPS_REVIEW"

    results.append({
        'invoice_id': invoice_id,
        'student_id': student_id,
        'payment_status': invoice['payment_status'],
        'batch_id': batch_id,
        'capacity_status': capacity_status,
        'reason_code': (
            ", ".join(sorted(set(issues)))
            if issues else "ALL_VALIDATIONS_PASSED"
        ),
        'decision': action
    })

# Final results
reconciliation_results = pd.DataFrame(results)

print("RECONCILIATION COMPLETED")

display(reconciliation_results)

print("\nDecision Summary:")

display(
    reconciliation_results['decision'].value_counts()
)
```

```python
# ==========================================
# CELL 9: GENERATE OPERATIONAL REPORTS
# ==========================================

# Separate eligible renewals
auto_approved = reconciliation_results[
    reconciliation_results['decision']
    == 'ELIGIBLE_FOR_AUTO_APPROVAL'
].copy()

# Separate exceptions
flagged_exceptions = reconciliation_results[
    reconciliation_results['decision']
    != 'ELIGIBLE_FOR_AUTO_APPROVAL'
].copy()

# Generate follow-up messages
def generate_message(row):

    reason = row['reason_code']
    student = row['student_id']
    invoice = row['invoice_id']

    if 'MISSING_FACULTY_FORM' in reason:
        return (
            f"Teacher: Please submit the renewal form "
            f"for Student {student}, Invoice {invoice}."
        )

    if 'DUPLICATE_FORM' in reason:
        return (
            f"Operations: Duplicate renewal form detected "
            f"for Student {student}. Please review."
        )

    if 'PAYMENT_NOT_COMPLETED' in reason:
        return (
            f"Parent: Payment for Invoice {invoice} "
            f"could not be verified. Please check payment status."
        )

    if 'AMOUNT_MISMATCH' in reason:
        return (
            f"Operations: Amount discrepancy found "
            f"for Invoice {invoice}. Please verify."
        )

    if 'ALREADY_SCHEDULED' in reason:
        return (
            f"Operations: Student {student} already has "
            f"an active schedule. No duplicate booking required."
        )

    if 'BOOKING_REQUEST_EXISTS' in reason:
        return (
            f"Operations: Existing booking request found "
            f"for Student {student}. Please check its status."
        )

    return (
        f"Operations: Please investigate Invoice {invoice}, "
        f"Student {student}. Reason: {reason}"
    )


followup_messages = flagged_exceptions.copy()

followup_messages['followup_message'] = (
    followup_messages.apply(generate_message, axis=1)
)

# Export three reports
auto_approved.to_csv(
    'auto_approved.csv', index=False
)

flagged_exceptions.to_csv(
    'flagged_exceptions.csv', index=False
)

followup_messages.to_csv(
    'followup_messages.csv', index=False
)

print("REPORTS GENERATED SUCCESSFULLY")

print("\nAuto Approval Records:", len(auto_approved))
print("Flagged Exceptions:", len(flagged_exceptions))
print("Follow-up Messages:", len(followup_messages))

print("\nAUTO APPROVED:")
display(auto_approved)

print("\nFLAGGED EXCEPTIONS:")
display(flagged_exceptions)

print("\nFOLLOW-UP MESSAGES:")
display(
    followup_messages[
        ['invoice_id', 'student_id', 'followup_message']
    ]
)
```

```python
# CELL 10: IMPROVE EXCEPTION CLASSIFICATION

# Create a copy of the reconciliation results
improved_results = reconciliation_results.copy()

# Identify existing booking request statuses
booking_status_map = (
    booking_requests
    .drop_duplicates(subset=['invoice_id'])
    .set_index('invoice_id')['request_status']
    .to_dict()
)

# Add workflow status
improved_results['workflow_status'] = (
    improved_results['invoice_id'].map(booking_status_map)
)

# Correctly identify invoice 84913's mismatched form reference
improved_results.loc[
    improved_results['invoice_id'] == 84913,
    'reason_code'
] = 'INVOICE_REFERENCE_MISMATCH'

# Classify existing booking requests
status_mapping = {
    'Completed': 'ALREADY_COMPLETED',
    'Pending_Batch_Allocation': 'PENDING_BATCH_ALLOCATION',
    'Failed_Batch_Full': 'BATCH_FULL_ALTERNATIVE_REQUIRED',
    'Awaiting_Parent_Batch_Selection': 'AWAITING_PARENT_SELECTION'
}

improved_results['workflow_status'] = (
    improved_results['workflow_status']
    .replace(status_mapping)
)

# Display improved results
display(
    improved_results[
        [
            'invoice_id',
            'student_id',
            'reason_code',
            'workflow_status',
            'decision'
        ]
    ]
)

print("\nImproved exception classification completed.")
```

```python
# ==========================================
# CELL 11: OPERATIONS DASHBOARD
# ==========================================

import matplotlib.pyplot as plt

# KPI calculations
total_invoices = len(improved_results)

auto_eligible = (
    improved_results['decision']
    == 'ELIGIBLE_FOR_AUTO_APPROVAL'
).sum()

manual_review = (
    improved_results['decision']
    == 'MANUAL_REVIEW'
).sum()

payment_issues = (
    improved_results['reason_code']
    .str.contains('PAYMENT', na=False)
).sum()

missing_forms = (
    improved_results['reason_code']
    .str.contains('MISSING_FACULTY_FORM', na=False)
).sum()

duplicate_forms = (
    improved_results['reason_code']
    .str.contains('DUPLICATE_FORM', na=False)
).sum()

print("========== OPERATIONS DASHBOARD ==========")

print("Total Invoices:", total_invoices)
print("Eligible for Auto Approval:", auto_eligible)
print("Manual Review:", manual_review)
print("Payment Issues:", payment_issues)
print("Missing Faculty Forms:", missing_forms)
print("Duplicate Forms:", duplicate_forms)

# Decision distribution chart
decision_counts = improved_results['decision'].value_counts()

plt.figure(figsize=(8, 5))
decision_counts.plot(kind='bar')
plt.title("Renewal Processing Decisions")
plt.xlabel("Decision")
plt.ylabel("Number of Invoices")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Workflow status chart
workflow_counts = (
    improved_results['workflow_status']
    .fillna('NEW / NOT YET BOOKED')
    .value_counts()
)

plt.figure(figsize=(10, 5))
workflow_counts.plot(kind='bar')
plt.title("Booking Workflow Status")
plt.xlabel("Workflow Status")
plt.ylabel("Number of Invoices")
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()

print("\nDashboard generated successfully.")
```

```python
# ==========================================
# CELL 12: AUTOMATION PERFORMANCE METRICS
# ==========================================

total = len(improved_results)

eligible = (
    improved_results['decision']
    == 'ELIGIBLE_FOR_AUTO_APPROVAL'
).sum()

manual = (
    improved_results['decision']
    == 'MANUAL_REVIEW'
).sum()

automation_rate = (eligible / total) * 100
manual_review_rate = (manual / total) * 100

metrics = pd.DataFrame({
    'Metric': [
        'Total Invoices',
        'Auto Approval Eligible',
        'Manual Review',
        'Automation Eligibility Rate (%)',
        'Manual Review Rate (%)'
    ],
    'Value': [
        total,
        eligible,
        manual,
        round(automation_rate, 2),
        round(manual_review_rate, 2)
    ]
})

display(metrics)

print("Automation Eligibility Rate:", round(automation_rate, 2), "%")
print("Manual Review Rate:", round(manual_review_rate, 2), "%")
```

```python
# ==========================================
# CELL 13: FINAL EXCEL WORKBOOK
# ==========================================

# Prepare final follow-up report using improved classifications

def create_final_message(row):

    reason = str(row['reason_code'])
    status = str(row['workflow_status'])
    invoice = row['invoice_id']
    student = row['student_id']

    if status == 'PENDING_BATCH_ALLOCATION':
        return "Operations: Allocate an appropriate batch."

    if status == 'BATCH_FULL_ALTERNATIVE_REQUIRED':
        return "Operations: Find an alternative batch for this student."

    if status == 'AWAITING_PARENT_SELECTION':
        return "Parent: Please select your preferred batch."

    if 'MISSING_FACULTY_FORM' in reason:
        return "Teacher: Please submit the missing renewal form."

    if 'INVOICE_REFERENCE_MISMATCH' in reason:
        return "Teacher: Please correct the invoice reference."

    if 'DUPLICATE_FORM' in reason:
        return "Operations: Review and resolve the duplicate form."

    if 'PAYMENT' in reason or 'AMOUNT_MISMATCH' in reason:
        return "Finance: Verify the payment status and amount."

    return f"Operations: Review Invoice {invoice}, Student {student}. Reason: {reason}"


# Exclude already completed bookings from actionable follow-ups
final_followups = improved_results[
    (improved_results['decision'] == 'MANUAL_REVIEW') &
    (improved_results['workflow_status'] != 'ALREADY_COMPLETED')
].copy()

final_followups['followup_message'] = (
    final_followups.apply(create_final_message, axis=1)
)

# Prepare dashboard metrics
dashboard_metrics = pd.DataFrame({
    'Metric': [
        'Total Invoices',
        'Auto Approval Eligible',
        'Manual Review',
        'Automation Eligibility Rate (%)',
        'Manual Review Rate (%)'
    ],
    'Value': [
        total,
        eligible,
        manual,
        round(automation_rate, 2),
        round(manual_review_rate, 2)
    ]
})

# Export all reports into one Excel workbook
file_name = 'Bambinos_Operations_Automation_Report.xlsx'

with pd.ExcelWriter(file_name, engine='openpyxl') as writer:

    auto_approved.to_excel(
        writer, sheet_name='Auto_Approved', index=False
    )

    flagged_exceptions.to_excel(
        writer, sheet_name='Flagged_Exceptions', index=False
    )

    final_followups.to_excel(
        writer, sheet_name='Followup_Messages', index=False
    )

    dashboard_metrics.to_excel(
        writer, sheet_name='Dashboard_Metrics', index=False
    )

    improved_results.to_excel(
        writer, sheet_name='All_Reconciliation', index=False
    )

    unique_batches.to_excel(
        writer, sheet_name='Batch_Capacity', index=False
    )

    # Format worksheets
    for worksheet in writer.sheets.values():

        worksheet.freeze_panes = 'A2'
        worksheet.auto_filter.ref = worksheet.dimensions

        for column_cells in worksheet.columns:
            max_length = max(
                len(str(cell.value or ''))
                for cell in column_cells
            )
            worksheet.column_dimensions[
                column_cells[0].column_letter
            ].width = min(max_length + 3, 45)

print("FINAL EXCEL WORKBOOK CREATED SUCCESSFULLY!")
print("File:", file_name)

# Download workbook to your computer
from google.colab import files
files.download(file_name)
```
